1. Never use abs(wins - losses) to represent record quality.

   It should be considered alongside total fight count.

3. Striking accuracy does not directly measure KO power.

4. Takedown frequency does not directly measure stamina.

5. Split-decision wins do not necessarily indicate low skill.

6. Raw KO and submission counts are influenced by career length.

7. Features that appear weak individually may become useful
   through interactions with other features.

## №4. Age Difference vs Win Rate

```python
age_diff = (train_df['R_age'] - train_df['B_age']).abs()

red_younger = (train_df['R_age'] < train_df['B_age'])
blue_younger = (train_df['B_age'] < train_df['R_age'])

younger_wins = (red_younger & (train_df['Winner'] == 'Red')) | ((blue_younger & (train_df['Winner'] == 'Blue')))

diff_1_2 = (age_diff >= 1) & (age_diff <=2)
diff_3_5 = (age_diff >= 3) & (age_diff <=5)
diff_6_8 = (age_diff >= 6) & (age_diff <=8)
diff_9_plus = age_diff >=9

total_1_2 = diff_1_2.sum()
total_3_5 = diff_3_5.sum()
total_6_8 = diff_6_8.sum()
total_9_plus = diff_9_plus.sum()

younger_wins_1_2 = (
    younger_wins & diff_1_2
).sum()
younger_wins_3_5 = (
    younger_wins & diff_3_5
).sum()
younger_wins_6_8 = (
    younger_wins & diff_6_8
).sum()
younger_wins_9_plus = (
    younger_wins & diff_9_plus
).sum()

winrate_1_2 = younger_wins_1_2 / total_1_2
winrate_3_5 = younger_wins_3_5 / total_3_5
winrate_6_8 = younger_wins_6_8 / total_6_8
winrate_9_plus = younger_wins_9_plus/ total_9_plus

group_names_young = ['Group 1-2', 'Group 3-5', 'Group 6-8', 'Group 9+']
total_winrate = [winrate_1_2, winrate_3_5, winrate_6_8, winrate_9_plus]

for name, winrate in zip(group_names_young, total_winrate):
    print(f'Group {name} winrate: {winrate}')
```

```python
total_fights = [total_1_2, total_3_5, total_6_8, total_9_plus]

younger_wins_count = [
    younger_wins_1_2,
    younger_wins_3_5,
    younger_wins_6_8,
    younger_wins_9_plus
]

for name, fights, wins, rate in zip(
    group_names_young,
    total_fights,
    younger_wins_count,
    total_winrate
):
    print(f"{name}: {fights} fights, {wins} wins, {rate:.2%}")
```
